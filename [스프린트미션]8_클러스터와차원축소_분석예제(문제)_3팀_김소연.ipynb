#### 이상치 제거 함수 정의

```python
# 이상치 제거 함수
def remove_out_outliers(df, col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    # 이상치를 제외한 데이터만 반환
    df = df[(df[col] >= lower) & (df[col] <= upper)]

    return df
```

### 문제16: DBSCAN 클러스터링 모델 학습 (매개변수 조정)
- DBSCAN의 eps와 min_samples 값을 조정하여, 클러스터링 결과가 개선되도록 하세요.

```python
from sklearn.neighbors import NearestNeighbors

# 데이터 X에 대해 min_samples와 동일한 k 값 설정
k = 5  # min_samples와 같은 값으로 설정
nearest_neighbors = NearestNeighbors(n_neighbors=k)
nearest_neighbors.fit(data5)
distances, indices = nearest_neighbors.kneighbors(data5)

# 각 포인트에서 k번째 최근접 이웃까지의 거리(즉, 5번째 이웃과의 거리) 추출
distances = np.sort(distances[:, k-1])

# 데이터의 95%에 해당하는 인덱스 계산
n = len(distances)  # 데이터 포인트의 총 개수
index_95 = int(n * 0.95)  # 95%에 해당하는 인덱스
index_70 = int(n * 0.7)  # 70%에 해당하는 인덱스

# 95%에 해당하는 y값
y_95 = distances[index_95]
y_70 = distances[index_70]

# K-거리 그래프 그리기
plt.plot(distances)
plt.title('K-distance Graph')
plt.xlabel('Data Points (sorted by distance)')
plt.ylabel(f'{k}th Nearest Neighbor Distance')
plt.axhline(y=y_70, color='g', linestyle='--', label=f'70% (y={y_70:.2f})')
plt.axhline(y=y_95, color='r', linestyle='--', label=f'95% (y={y_95:.2f})')
plt.legend()
plt.show()
```

### 문제22: 적절한 주성분 수 결정
- 적절한 주성분 수를 결정하기 위해, 주성분의 설명력(분산 설명 비율)을 시각화하세요.

```python
pca = PCA()
pca.fit(data7)

explained_variance_ratio = pca.explained_variance_ratio_  # 각 주성분의 설명력
cumulative_variance_ratio = explained_variance_ratio.cumsum()  # 누적 설명력

# 시각화
plt.figure(figsize=(12, 8))
sns.lineplot(x=range(1, len(explained_variance_ratio) + 1), y=explained_variance_ratio,
             marker='o', color='blue', label='개별 설명 분산')
sns.lineplot(x=range(1, len(cumulative_variance_ratio) + 1), y=cumulative_variance_ratio,
             marker='o', linestyle='-', color='red', label='누적 설명 분산')

# 누적합이 70%에 도달하는 지점에 수직 파선 추가
threshold_index = np.argmax(cumulative_variance_ratio >= 0.7)  # 누적 설명력이 70% 이상이 되는 첫 번째 인덱스
plt.axvline(x=threshold_index + 1, color='green', linestyle='--', label=f"PC{threshold_index + 1} (70%)")

# 축 및 제목 설정
plt.xlabel('주성분 번호', fontsize=12)
plt.ylabel('설명 분산 비율', fontsize=12)
plt.title('설명 분산 비율과 누적 설명 분산', fontsize=16)
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.lineplot(x=range(1, 18), y=explained_variance_ratio, marker='o', color='blue')
plt.xlabel('Number of Principal Components')
plt.ylabel('Variance Explained')
plt.title('PCA Explained Variance Ratio')
plt.grid(True)
plt.show()
```

### 문제27 : PCA를 활용한 차원 축소 후 DBSCAN 클러스터링
- PCA를 통해 차원 축소한 데이터를 사용하여 DBSCAN 클러스터링을 수행하고 결과를 시각화하세요.

```python
# pca_df = pca_df.drop(columns='DBSCAN_Cluster')
pca_df
```

```python
# pca_df에 대한 K-최근접 이웃 계산
# k = 5로 설정 (min_samples와 동일한 값으로 설정)
k = 5

# NearestNeighbors 모델 설정
nearest_neighbors = NearestNeighbors(n_neighbors=k)
nearest_neighbors.fit(pca_df)  # pca_df 데이터에 대해 K-NN 계산

# K-최근접 이웃 거리 계산
distances, indices = nearest_neighbors.kneighbors(pca_df)

# 각 포인트에서 k번째 최근접 이웃까지의 거리(즉, 5번째 이웃과의 거리) 추출
distances = np.sort(distances[:, k-1])

# 데이터의 95%와 70%에 해당하는 인덱스 계산
n = len(distances)  # 데이터 포인트의 총 개수
index_95 = int(n * 0.95)  # 95%에 해당하는 인덱스
index_70 = int(n * 0.7)  # 70%에 해당하는 인덱스

# 95% 및 70%에 해당하는 y값 추출
y_95 = distances[index_95]
y_70 = distances[index_70]

# K-거리 그래프 그리기
plt.plot(distances)
plt.title('K-distance Graph')
plt.xlabel('Data Points (sorted by distance)')
plt.ylabel(f'{k}th Nearest Neighbor Distance')
plt.axhline(y=y_70, color='g', linestyle='--', label=f'70% (y={y_70:.2f})')
plt.axhline(y=y_95, color='r', linestyle='--', label=f'95% (y={y_95:.2f})')
plt.grid(True)
plt.legend()
plt.show()

# DBSCAN 모델 학습 (eps 값은 95%에 해당하는 y_95로 설정)
dbscan = DBSCAN(eps=y_95, min_samples=k)  # eps: 95% 지점, min_samples: k 값
dbscan.fit(pca_df)

# 클러스터 결과를 pca_df에 추가
pca_df['DBSCAN_Cluster'] = dbscan.labels_
```

```python
# DBSCAN 모델 학습 (eps=0.86, min_samples=5)
eps = 0.83  # K-거리 그래프에서 도출된 95%
k = 5

dbscan = DBSCAN(eps=eps, min_samples=k)
dbscan.fit(pca_df)

# DBSCAN 클러스터 결과를 추가
pca_df['DBSCAN_Cluster'] = dbscan.labels_

# DBSCAN 클러스터 레이블 확인
print("DBSCAN 클러스터 레이블:", np.unique(dbscan.labels_))

# 클러스터링 결과 확인
print("DBSCAN 클러스터링 완료.")
```

K-means는 주어진 클러스터 수에 맞춰 데이터를 나누는 알고리즘으로, 차원 축소 후 클러스터 간의 경계가 더 명확해지기 때문에 해석이 용이해진다.
그러나 클러스터 수를 미리 지정해야 하므로, 적절한 클러스터 수를 선정하는 것이 중요한 점으로 작용한다.

2. Hierarchical Clustering (계층적 클러스터링)  
계층적 클러스터링은 이전 후 결과가 잘 이해되지 않는다는 의견이 많다. 특히, purchase와 balance 같은 데이터의 산점도에는 적합하지 않다는 결론을 내릴 수 있다.
계층적 클러스터링은 클러스터의 수를 미리 정할 수 없으며, 트리 구조를 생성하는 특성 때문에 데이터가 복잡할 경우 해석이 어려워질 수 있다.

3. DBSCAN (Density-Based Spatial Clustering of Applications with Noise)  
DBSCAN은 밀도 기반 클러스터링으로, 클러스터 간 경계가 불분명하거나 이상치가 많은 경우에 유리하다.  
차원 축소 후 밀도가 높은 영역을 잘 파악할 수 있기 때문에, 더 적합한 클러스터를 형성할 수 있다.  
차원 축소 전에는 닭다리 모양 분포 근처에서 이상치가 발견되었지만, 차원 축소 후에는 해당 이상치들이 클러스터의 내부로 들어오는 현상이 발생한 이유는 차원 축소가 이상치의 영향을 줄여서 일 수도 있을 것이다.(추가적으로 알아봐야할 사항)

4. GMM (Gaussian Mixture Model)  
GMM은 가우시안 분포를 따르는 클러스터링 기법으로, 각 클러스터가 정규분포를 따른다고 가정한다.
이 모델은 클러스터 모양이 원형에 가까운 경우 적합하며, 확률적인 접근으로 더 유연한 클러스터링을 제공한다.  
그러나 GMM은 클러스터 수와 분포 모형에 대한 가정이 필요하기 때문에, 데이터가 정규분포를 따르지 않으면 성능이 떨어질 수 있다.

5. 한계  
차원 축소와 클러스터링 기법은 데이터 특성에 따라 다르게 작용할 수 있다.  
복잡한 데이터나 클러스터 간 경계가 명확하지 않은 경우, 차원 축소와 클러스터링 기법이 잘못된 클러스터를 형성할 위험이 있다.  
각 기법의 적합성은 데이터의 분포와 특성에 맞추어 신중하게 선택해야 한다.
특히, 클러스터가 불규칙한 모양을 가지거나 이상치가 많은 경우 DBSCAN이나 GMM이 유리할 수 있다.

6. 결론  
각 클러스터링 기법은 데이터의 특성에 따라 다르게 작용하며, 상황에 맞는 기법을 선택하는 것이 중요하다.  
특히 클러스터 경계가 불명확한 경우에는 DBSCAN이나 GMM이 더 유리할 수 있으며, 차원 축소가 기법의 성능을 개선하거나 해석을 용이하게 만들 수 있다.
